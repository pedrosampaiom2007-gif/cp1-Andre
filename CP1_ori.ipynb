Uma equipe responsável pelo planejamento da distribuição de energia em Tétouan deseja identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda.

```python
df3.info()
```

```python
df3.describe()
```

```python
#Renomear nome das variáveis de consumo
df3 = df3.rename(columns={
    'Zone 1 Power Consumption': 'Consumo_Zona_1',
    'Zone 2  Power Consumption': 'Consumo_Zona_2',
    'Zone 3  Power Consumption': 'Consumo_Zona_3',
})
df3
```

```python
max_zona1 = df3["Consumo_Zona_1"].max()
max_zona2 = df3["Consumo_Zona_2"].max()
max_zona3 = df3["Consumo_Zona_3"].max()

print(f"Zona 1: {max_zona1:.2f}")
print(f"Zona 2: {max_zona2:.2f}")
print(f"Zona 3: {max_zona3:.2f}")
```

```python
picos = {
    "Consumo_Zona_1": max_zona1,
    "Consumo_Zona_2": max_zona2,
    "Consumo_Zona_3": max_zona3
}

zona_pico = max(picos, key=picos.get)
PMAX = picos[zona_pico]

print(f"A zona com maior pico de consumo é {zona_pico}, com valor de {PMAX:.2f}")
```

A zona com maior pico de consumo na amostra foi a Consumo_Zona_1, com valor máximo de 51571.50. Isso indica que essa é a região com maior demanda de energia entre os momentos registrados.

```python
P70 = 0.70 * PMAX
print(f"Limiar (70% do pico): {P70:.2f}")

df_alto = df3[df3[zona_pico] > P70]
df_alto.head()
```

```python
qtd_alto = len(df_alto)
qtd_total = len(df3)
percentual_alto = qtd_alto / qtd_total * 100

print(f"Registros acima do limiar: {qtd_alto} de {qtd_total}")
print(f"Percentual: {percentual_alto:.2f}%")
```

```python
# Segundo criterio = temperatura AMBIENTE (variavel Temperature), nao o consumo da zona -
# e essa condicao ambiental que o enunciado pede, no mesmo padrao dos exercicios 1 e 6.
temperatura_media = df3['Temperature'].mean()
print(f"Temperatura ambiente média: {temperatura_media:.2f}")

df_alto_temp = df3[
    (df3[zona_pico] > P70) & (df3['Temperature'] > temperatura_media)
]

qtd_alto_temp = len(df_alto_temp)
percentual_alto_temp = qtd_alto_temp / qtd_total * 100

print(f"Registros com consumo acima do limiar E temperatura ambiente acima da média: {qtd_alto_temp}")
print(f"Percentual: {percentual_alto_temp:.2f}%")
```

```python
print("Comparação entre os dois DataFrames:")
print(f"Somente consumo acima do limiar:        {qtd_alto} registros ({percentual_alto:.2f}%)")
print(f"Consumo acima do limiar + temp. acima:  {qtd_alto_temp} registros ({percentual_alto_temp:.2f}%)")
print(f"Diferença: {qtd_alto - qtd_alto_temp} registros a menos")
```

```python
if percentual_solar > percentual_eolica:
    fonte_mais_frequente = "solar"
elif percentual_eolica > percentual_solar:
    fonte_mais_frequente = "eolica"
else:
    fonte_mais_frequente = "as duas fontes empataram"

print(f"A fonte que aparece com maior frequencia acima de 70% do seu proprio maximo e a {fonte_mais_frequente}.")
```

## Dataset 6 — Individual Household Electric Power Consumption (UCI)
LINK: https://www.kaggle.com/datasets/anikannal/solar-power-generation-data

Situação

Uma residência possui monitoramento elétrico detalhado e deseja identificar episódios de demanda elevada que também apresentem corrente acima do comportamento médio. O dataset é o mesmo utilizado como referência em aula, mas o critério de análise será diferente.

```python
import pandas as pd

df6 = pd.read_csv("SAMPLE_ENERGY_DATA.csv")
```

```python
# RENOMEAR
df6 = df6.rename(columns={
"Global_active_power": "Pot_Ativa",
"Global_reactive_power": "Pot_Reativa",
"Voltage": "Tensao",
"Global_intensity": "Corrente",
"Sub_metering_1": "Sub1",
"Sub_metering_2": "Sub2",
"Sub_metering_3": "Sub3"
})

df6.head()
df6.dtypes
```

```python
# MAXIMO

PMAX = df6["Pot_Ativa"].max()
print(f"A potência ativa máxima é de: {PMAX}")
```

```python
# 75%

P75 = 0.75 * PMAX
print(f"75% da potência ativa é de: {P75}")
```

```python
# DataFrame acima do limite

df6_alto = df6[df6["Pot_Ativa"] > P75]
df6_alto.head()
```

```python
# Quantidade e percentual

qtd_alto = df6_alto.shape[0]
qtd_total = df6.shape[0]
percentual_alto = qtd_alto / qtd_total * 100

print(f"Registros com potência acima de 75% do máximo: {qtd_alto}")
print(f"Total de registros na amostra: {qtd_total}")
print(f"Percentual: {percentual_alto:.2f}%")
```

```python
# Corrente média

corrente_media = df6['Corrente'].mean()
print(f"Corrente média: {corrente_media:.2f} A")
```

```python
# Segundo DataFrame

df6_alto_corrente = df6[(df6['Pot_Ativa'] > P75) & (df6['Corrente'] > corrente_media)]
df6_alto_corrente.head()
```

```python
qtd_alto_corrente = df6_alto_corrente.shape[0]
percentual_alto_corrente = qtd_alto_corrente / qtd_total * 100

print(f"Registros com potência alta E corrente acima da média: {qtd_alto_corrente}")
print(f"Percentual: {percentual_alto_corrente:.2f}%")
```

```python
# Define df6_alto (originally from cell c9gSL2uPHlXq), assuming df6 and P75 are defined in prior cells.
df6_alto = df6[df6["Pot_Ativa"] > P75]

qtd_alto = df6_alto.shape[0]
qtd_total = df6.shape[0]
percentual_alto = qtd_alto / qtd_total * 100

# Comparação

print(f"Registros somente com potência alta: {qtd_alto} ({percentual_alto:.2f}%)")
print(f"Registros com potência alta E corrente acima da média: {qtd_alto_corrente} ({percentual_alto_corrente:.2f}%)")

print("\nPotência ativa e corrente estão fortemente relacionadas (P ≈ V x I, com a tensao praticamente constante). Por isso, quando o critério de corrente acima da média é somado ao de potência alta, o conjunto muda pouco ou nada: os registros de pico de potência já tendem a ter corrente bem acima da média da amostra, entao a segunda condição normalmente nao filtra registros adicionais, apenas confirma o que a potência já indicava.")
```

## Comparação

A adição da condição de corrente acima da média não alterou o conjunto resultante: ambos os DataFrames possuem 36 registros (0.04% da amostra). Isso ocorre porque potência ativa e corrente são fortemente correlacionadas (P ≈ V × I, com tensão praticamente constante). Como a corrente média da amostra (4.64 A) é muito inferior às correntes dos registros de pico de potência (33-36 A), todos os registros que já atendiam ao critério de potência elevada também atendiam ao critério de corrente acima da média, tornando a segunda condição redundante nesse caso.
